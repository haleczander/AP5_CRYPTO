Using the knowledge of $e'$ and $d'$, launch a Fermat attack against $n$ in order to recover its prime factors $p$ and $q$.

```python
import math, random

def factor_from_ed(e, d, n, tries=50):

    k = (e * d) - 1
    if k % 2 == 1:
        raise ValueError("k doit être pair")
    
    s = 0
    r = k
    while r % 2 == 0:
        r //= 2
        s += 1

    for attempt in range(tries):
        g = random.randrange(2, n-1)
        y = pow(g, r, n)
        if y == 1 or y == n-1:
            continue
        for j in range(s):
            x = pow(y, 2**j, n)
            if x == 1:
                p = math.gcd(y-1, n)
                if 1 < p < n:
                    q = n // p
                    return p, q
                break
            if x == n-1:
                break
    raise RuntimeError("Échec : augmenter 'tries' ou vérifier les entrées")
```

```python
p,q=factor_from_ed(ee, dd, n)
print("p", hex(p))
print("q", hex(q))
print("n", p*q == n )
```

## D) Generating keys

Generating fresh parameters $(n,e,d)$ to use in RSA is, however, quite simple (_i.e._, reasonably fast). Help Alice and Bob by generating

- two 1536-bit prime factors $p$ and $q$;
- the corresponding RSA modulus $n$;
- a public exponent $e$ of your liking;
- the corresponding private exponent $d$.

For the first step, you can use the provided `is_prime()` function to quickly test if a given integer is (with large probability) prime. How long does generating RSA keys take on your machine? You can compare the performance of your key generation algorithm with that of a standard library, _e.g._ [OpenSSL](https://docs.openssl.org/3.1/man1/openssl-genrsa/).

```python
import random
from is_prime import is_prime

def generate_prime(bits):
    while True:
        p = random.getrandbits(bits) | 1  # s'assurer que p est impair
        if is_prime(p):
            return p
        
def generate_rsa(e, bits):
    p, q = generate_prime(bits), generate_prime(bits)
    n = p * q
    d = calc_d(e,p,q)
    return n, d
```

```python
def rsa_encrypt(message: bytes, n, e):
    m = int.from_bytes(message, "big")
    c = pow(m, e, n)
    return c

def rsa_decrypt(ciphertext: int, n, d):
    m = pow(ciphertext, d, n)
    return m.to_bytes((m.bit_length() + 7) // 8, "big")
```

```python
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.backends import default_backend

def generate_rsa_hazmat(bits=1536, e=65537):
    # Génère la clé privée RSA
    private_key = rsa.generate_private_key(
        public_exponent=e,
        key_size=bits * 2,        #  modulus ~2*1536 bits
        backend=default_backend()
    )
    
    numbers = private_key.private_numbers()
    p = numbers.p
    q = numbers.q
    d = numbers.d
    n = numbers.public_numbers.n
    e = numbers.public_numbers.e

    return n, e, d
```

```python
from statistics import mean
from time import time

def benchmark(func, label, loops=10):
    times = []
    for _ in range(loops):
        t0 = time()
        func()
        times.append(time() - t0)
    print(f"\n{label}")
    print(f"  → min :  {min(times):.4f}s")
    print(f"  → max :  {max(times):.4f}s")
    print(f"  → mean : {mean(times):.4f}s")

bits = 1536
e = 65537

benchmark(lambda: generate_rsa_hazmat(bits, e), "Hazmat")
benchmark(lambda: generate_rsa(e, bits), "Maison")
```
