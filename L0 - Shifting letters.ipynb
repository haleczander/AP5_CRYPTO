### Shifting letters

```python
print("Yo man")
```

```python
# run this first

class S( object ):
    
    def __init__(self, s):
    
        if type(s) == str:
            self.s = s
        elif type(s) == int:
            self.s = chr(s + 64)
        elif type(s) == list:
            self.s = "".join( chr(c + 64) for c in s )
        elif type(s) == S:
            self.s = s.s
        
    def __repr__(self):
        
        return self.s
        
    def list(self):
        
        return [ ord(c) - 64 for c in self.s ]
        
    def __add__(self,other):
        
        oth = S(other)
        q, r = len(self.s)//len(oth.s), len(self.s) % len(oth.s)
        return S([ (c + d - 1) % 26 + 1 for (c,d) in zip(self.list(), q*oth.list() + oth.list()[:r] )])
        
    def __neg__(self):
        
        return S([ (-c - 1) % 26 + 1 for c in self.list() ])
        
    def __sub__(self, other):
        
        return self + (-other)
    
    def __len__(self):
        
        return len(self.s)
    
print("Shifting library loaded")
```

### Ceasar's cipher

Apply a uniform shift to all letters in the plaintext.

```python
m = S("HELLO")
m

```

```python
k = 3
c = m + k
c
```

To decrypt, just apply the inverse shift to the ciphertext.

```python
c - k
```

### Brute-force attack

This cryptosystem is _very_ insecure as it's easy to try all the possible keys ($\sim 5$ bit security level).

```python
for k in range(26):
    print(k, c - k)
```

###### It's actually much worse...

$\sim 0$ bit security level!

```python
c = S("AOLJHLZHYJPWOLYPZUVACLYFZLJBYL")

k = 1  # you have only 1 chance

c - k
```

### Solution: use a longer key

```python
from random import randrange

def randkey(n):
    return S([ randrange(1,27) for i in range(1,n+1) ])
```

```python
m = S("ENCRYPTASTRINGBYRANDOMLYSHIFTINGEVERYLETTER")

k = randkey(len(m))

print("plaintext:  ", m)
print("key:        ", k)
print("ciphertext: ", m + k)
```

Extreme cases:

- $\mathrm{len}(k) = 1$: Ceasar
- 1 < $\mathrm{len}(k) < \mathrm{len}(m)$: Vigenère
- $\mathrm{len}(k) = \mathrm{len}(m)$: Vernam (more next week)
